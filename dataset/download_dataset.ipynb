```python
import kagglehub
import shutil
import os

# Step 1: Download dataset to default path
download_path = kagglehub.dataset_download("zalando-research/fashionmnist")

print("Downloaded to:", download_path)

# Step 2: Define new target path
new_path = "/Users/sairameti/Documents/fashion-classification-using-CNN/dataset"

# Step 3: Create target directory if it doesn't exist
os.makedirs(new_path, exist_ok=True)

# Step 4: Move the entire folder
shutil.move(download_path, new_path)

print(f"✅ Dataset moved to: {os.path.join(new_path, os.path.basename(download_path))}")
```
